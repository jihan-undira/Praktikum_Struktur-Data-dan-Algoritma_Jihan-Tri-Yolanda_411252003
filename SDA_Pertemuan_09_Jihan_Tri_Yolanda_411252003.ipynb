```python
print('Lingkungan Pertemuan 9 siap')
from collections import deque
print('deque berhasil diimpor')

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

print('Class Node berhasil didefinisikan')
```

# Praktik Mandiri 9.1, 9.2, & 9.3

```python
# Membangun tree awal A - G
root = Node('A')
root.left = Node('B')
root.right = Node('C')
root.left.left = Node('D')
root.left.right = Node('E')
root.right.left = Node('F')
root.right.right = Node('G')

# --- Praktik Mandiri 9.1 ---
root.left.left.left = Node('H')
print("=== Praktik Mandiri 9.1 ===")
print("Data node baru (root.left.left.left.data):", root.left.left.left.data)

# --- Praktik Mandiri 9.2 ---
def inorder(node):
    if node is None:
        return
    inorder(node.left)
    print(node.data, end=' ')
    inorder(node.right)

print("\n=== Praktik Mandiri 9.2 ===")
print("Hasil Inorder:", end=' ')
inorder(root)
print()

# --- Praktik Mandiri 9.3 ---
def count_nodes(node):
    if node is None:
        return 0
    return 1 + count_nodes(node.left) + count_nodes(node.right)

def count_leaves(node):
    if node is None:
        return 0
    if node.left is None and node.right is None:
        return 1
    return count_leaves(node.left) + count_leaves(node.right)

print("\n=== Praktik Mandiri 9.3 ===")
print("Jumlah node:", count_nodes(root))
print("Jumlah leaf:", count_leaves(root))
```

# Praktik Mandiri 9.4 (Membangun Tree Sendiri K-D-M-B-F-Z)

```python
# Membangun tree K
root_k = Node('K')
root_k.left = Node('D')
root_k.right = Node('M')
root_k.left.left = Node('B')
root_k.left.right = Node('F')
root_k.right.right = Node('Z')

# Fungsi-fungsi traversal, height, dan search
def preorder_list(node):
    if node is None:
        return []
    return [node.data] + preorder_list(node.left) + preorder_list(node.right)

def inorder_list(node):
    if node is None:
        return []
    return inorder_list(node.left) + [node.data] + inorder_list(node.right)

def postorder_list(node):
    if node is None:
        return []
    return postorder_list(node.left) + postorder_list(node.right) + [node.data]

def level_order(root):
    if root is None:
        return []
    result = []
    queue = deque([root])
    while queue:
        node = queue.popleft()
        result.append(node.data)
        if node.left is not None:
            queue.append(node.left)
        if node.right is not None:
            queue.append(node.right)
    return result

def height(node):
    if node is None:
        return -1
    return 1 + max(height(node.left), height(node.right))

def search(node, target):
    if node is None:
        return False
    if node.data == target:
        return True
    return search(node.left, target) or search(node.right, target)

print("=== Praktik Mandiri 9.4 ===")
print("Preorder   :", preorder_list(root_k))
print("Inorder    :", inorder_list(root_k))
print("Postorder  :", postorder_list(root_k))
print("Level-order:", level_order(root_k))
print("Jumlah node:", count_nodes(root_k))
print("Jumlah leaf:", count_leaves(root_k))
print("Height tree:", height(root_k))
print("Cari 'F'   :", search(root_k, 'F'))
print("Cari 'X'   :", search(root_k, 'X'))
```

# Praktik Mandiri 9.5 (Simulasi Trace Call Stack inorder)

```python
def inorder_trace(node, depth=0):
    indent = "  " * depth
    if node is None:
        print(f"{indent}-> inorder(None): base case, return")
        return
    print(f"{indent}-> Masuk inorder({node.data})")
    inorder_trace(node.left, depth + 1)
    print(f"{indent}   [CETAK: {node.data}]")
    inorder_trace(node.right, depth + 1)
    print(f"{indent}<- Keluar inorder({node.data})")

# Gunakan tree A-G standar tanpa H untuk trace Praktik 9.5
root_ag = Node('A')
root_ag.left = Node('B')
root_ag.right = Node('C')
root_ag.left.left = Node('D')
root_ag.left.right = Node('E')
root_ag.right.left = Node('F')
root_ag.right.right = Node('G')

print("=== Praktik Mandiri 9.5 (Trace Call Stack) ===")
inorder_trace(root_ag)
```

# Praktik Mandiri 9.6 (Perbandingan Recursive vs Iterative)

```python
def preorder_iterative(root):
    if root is None:
        return []
    result = []
    stack = [root]
    while stack:
        node = stack.pop()
        result.append(node.data)
        if node.right is not None:
            stack.append(node.right)
        if node.left is not None:
            stack.append(node.left)
    return result

print("=== Praktik Mandiri 9.6 ===")
assert preorder_list(root_ag) == preorder_iterative(root_ag)
print("Hasil sama:", preorder_iterative(root_ag))
```

# Latihan Evaluasi Nomor 10, 11, & 12

```python
# --- Nomor 10: Menjumlahkan seluruh nilai numerik pada binary tree ---
def sum_tree(node):
    if node is None:
        return 0
    return node.data + sum_tree(node.left) + sum_tree(node.right)

# --- Nomor 11: Menghitung jumlah internal node (minimal 1 child) ---
def count_internal(node):
    if node is None:
        return 0
    if node.left is None and node.right is None:
        return 0
    return 1 + count_internal(node.left) + count_internal(node.right)

# --- Nomor 12: Expression Tree dengan operator % dan ** ---
def evaluate(node):
    if node.left is None and node.right is None:
        return node.data

    left_val = evaluate(node.left)
    right_val = evaluate(node.right)

    if node.data == '+':
        return left_val + right_val
    elif node.data == '-':
        return left_val - right_val
    elif node.data == '*':
        return left_val * right_val
    elif node.data == '/':
        return left_val / right_val
    elif node.data == '%':
        return left_val % right_val
    elif node.data == '**':
        return left_val ** right_val
    else:
        raise ValueError(f"Operator tidak dikenali: {node.data}")

# Pengujian Nomor 10 (Tree angka)
num_root = Node(10)
num_root.left = Node(5)
num_root.right = Node(15)
num_root.left.left = Node(2)
num_root.left.right = Node(3)

print("=== Latihan Evaluasi No. 10 & 11 ===")
print("Total sum_tree      :", sum_tree(num_root))        # Output: 35
print("Jumlah internal node:", count_internal(num_root))  # Output: 2 (node 10 dan 5)

# Pengujian Nomor 12: Ekspresi (2 ** 3) + (10 % 3)
expr_test = Node('+')
expr_test.left = Node('**')
expr_test.right = Node('%')
expr_test.left.left = Node(2)
expr_test.left.right = Node(3)
expr_test.right.left = Node(10)
expr_test.right.right = Node(3)

print("\n=== Latihan Evaluasi No. 12 ===")
print("Hasil evaluasi (2 ** 3) + (10 % 3):", evaluate(expr_test))  # Output: 9
```
