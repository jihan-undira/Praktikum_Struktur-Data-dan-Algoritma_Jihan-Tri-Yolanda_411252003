## 2.1 Operasi Array
Menguji pembuatan list, pengaksesan elemen, pembaruan data, penambahan, penyisipan, penghapusan, serta traversal pada array/list Python.

### 1. Inisialisasi List
Membuat sebuah list bernama `nilai` yang berisi 10 nilai integer.

```python
nilai = [10, 25, 30, 45, 50, 65, 70, 85, 90, 100]
print("List awal:", nilai)
```

### 2. Pengaksesan Elemen
Menampilkan elemen pertama (indeks 0), elemen ke-5 (indeks 4), dan elemen terakhir (indeks -1).

```python
print("Elemen pertama (indeks 0):", nilai[0])
print("Elemen ke-5 (indeks 4):", nilai[4])
print("Elemen terakhir (indeks -1):", nilai[-1])
```

### 3. Memperbarui Nilai Elemen

### 8. Komentar Operasi yang Menyebabkan Pergeseran Elemen
**Analisis Pergeseran Elemen (*Element Shift*):**
1. **`insert(index, val)`**: Menyebabkan pergeseran elemen dari posisi `index` ke kanan sebesar satu langkah untuk memberi ruang bagi nilai baru.
2. **`pop(index)`**: Menyebabkan pergeseran elemen setelah `index` ke kiri sebesar satu langkah untuk mengisi posisi yang kosong.
3. **`remove(val)`**: Mencari lokasi elemen `val` terlebih dahulu, lalu menggeser seluruh elemen di sebelahnya ke kiri setelah elemen tersebut dihapus.

## 2.2 Linear Search
Menguji algoritma pencarian linier dengan menelusuri data secara manual satu per satu tanpa menggunakan operator `in` atau method bawaan.

### 1. Definisi Fungsi Linear Search dengan Trace
Membuat fungsi pencarian linier buatan sendiri yang menghitung dan mengembalikan indeks target beserta jumlah pemeriksaan data yang dilakukan.

```python
def linear_search_trace(arr, target):
    jumlah_pemeriksaan = 0
    for idx, val in enumerate(arr):
        jumlah_pemeriksaan += 1

        # Penanganan opsional agar perbandingan string tidak membedakan huruf besar/kecil (case-insensitive)
        val_cmp = val.lower() if isinstance(val, str) else val
        target_cmp = target.lower() if isinstance(target, str) else target

        if val_cmp == target_cmp:
            return idx, jumlah_pemeriksaan

    return -1, jumlah_pemeriksaan

# Data list minimal 15 bilangan tidak terurut
data_linear = [42, 15, 88, 23, 7, 91, 34, 56, 12, 67, 80, 5, 49, 73, 29]
print("Data awal (15 elemen):", data_linear)
```

### 2. Pengujian Best Case (Target di Elemen Pertama)
Mencari nilai 42 yang berada di posisi indeks 0 dan mencatat jumlah pemeriksaannya.

```python
idx1, count1 = linear_search_trace(data_linear, 42)
print(f"Target: 42 | Indeks Ditemukan: {idx1} | Jumlah Pemeriksaan: {count1}")
```

### 3. Pengujian Worst Case 1 (Target di Elemen Terakhir)
Mencari nilai 29 yang berada di posisi indeks terakhir dan mencatat jumlah pemeriksaannya.

```python
idx2, count2 = linear_search_trace(data_linear, 29)
print(f"Target: 29 | Indeks Ditemukan: {idx2} | Jumlah Pemeriksaan: {count2}")
```

### 4. Pengujian Worst Case 2 (Target Tidak Ditemukan)
Mencari nilai 999 yang tidak ada di dalam list data dan mencatat jumlah pemeriksaannya.

```python
idx3, count3 = linear_search_trace(data_linear, 999)
print(f"Target: 999 | Indeks Ditemukan: {idx3} | Jumlah Pemeriksaan: {count3}")
```

### 5. Analisis Kasus Best Case dan Worst Case
* **Best Case**: Terjadi pada pengujian target `42` (elemen pertama) dengan hanya **1 kali pemeriksaan**.
* **Worst Case**: Terjadi pada target `29` (elemen terakhir) dan `999` (tidak ada di data) dengan **15 kali pemeriksaan** (sama dengan jumlah total elemen $n$).

### 6. Tantangan Opsional: Case-Insensitive Linear Search untuk Teks
Menguji modifikasi Linear Search menggunakan method `.lower()` agar pencarian teks tidak terpengaruh huruf besar/kecil.

```python
data_nama = ["Budi", "Andi", "Cici", "Deni"]
target_nama = "andi"

idx_nama, count_nama = linear_search_trace(data_nama, target_nama)
print(f"Cari '{target_nama}' pada {data_nama}")
print(f"Hasil -> Indeks: {idx_nama}, Jumlah Pemeriksaan: {count_nama}")
```

### 7. Penjelasan Mengapa Worst Case Linear Search adalah O(n)
Linear Search bekerja dengan cara memeriksa setiap elemen list satu per satu secara sekuensial dari awal hingga akhir. Dalam skenario terburuk (*worst case*), yaitu ketika elemen berada di paling akhir atau tidak ada sama sekali, algoritma terpaksa membandingkan target dengan seluruh elemen yang berjumlah $n$. Karena jumlah iterasi dan pemeriksaan meningkat secara sebanding lurus dengan banyaknya elemen, maka kompleksitas waktunya dinyatakan sebagai $O(n)$.

## 2.3 Binary Search
Menguji algoritma pencarian biner pada data terurut dengan melacak pergerakan variabel `left`, `right`, dan `mid`.

### 1. Persiapan Data Terurut
Membuat list minimal 20 angka acak dan mengurutkannya menggunakan `sorted()`.

```python
data_raw = [88, 12, 45, 67, 23, 90, 34, 11, 78, 56, 3, 99, 14, 25, 61, 82, 40, 71, 19, 50]
data_sorted = sorted(data_raw)
print("Data Terurut (20 elemen):", data_sorted)
```

### 2. Fungsi Binary Search dengan Penelusuran (Trace)
Fungsi ini mencatat rentang pencarian `left`, `right`, dan posisi `mid` pada setiap iterasi.

```python
def binary_search_trace(arr, target):
    left = 0
    right = len(arr) - 1
    jumlah_pemeriksaan = 0
    trace_table = []

    while left <= right:
        jumlah_pemeriksaan += 1
        mid = (left + right) // 2
        trace_table.append((jumlah_pemeriksaan, left, right, mid, arr[mid]))

        if arr[mid] == target:
            return mid, jumlah_pemeriksaan, trace_table
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1, jumlah_pemeriksaan, trace_table
```

### 3. Pengujian Target di Sekitar Tengah, Awal, Akhir, dan Tidak Ada
Jalankan pencarian biner untuk beberapa skenario posisi target.

```python
# Target di sekitar tengah (45)
idx_mid, count_mid, trace_mid = binary_search_trace(data_sorted, 45)

# Target dekat awal (11)
idx_low, count_low, trace_low = binary_search_trace(data_sorted, 11)

# Target dekat akhir (90)
idx_high, count_high, trace_high = binary_search_trace(data_sorted, 90)

# Target tidak ada (100)
idx_none, count_none_bs, trace_none = binary_search_trace(data_sorted, 100)

print(f"Target 45 (Tengah) -> Indeks: {idx_mid}, Jumlah Pemeriksaan: {count_mid}")
print(f"Target 11 (Awal)   -> Indeks: {idx_low}, Jumlah Pemeriksaan: {count_low}")
print(f"Target 90 (Akhir)  -> Indeks: {idx_high}, Jumlah Pemeriksaan: {count_high}")
print(f"Target 100 (Tidak Ada) -> Indeks: {idx_none}, Jumlah Pemeriksaan: {count_none_bs}")
```

### 4. Tabel Penelusuran (Trace Table)
Menampilkan tabel nilai `left`, `right`, dan `mid` pada setiap iterasi untuk pencarian target 90.

```python
print("Tabel Trace Binary Search (Target: 90):")
print(f"{'Iterasi':<8} | {'Left':<6} | {'Right':<6} | {'Mid':<6} | {'Nilai Mid':<10}")
print("-" * 45)
for item in trace_high:
    print(f"{item[0]:<8} | {item[1]:<6} | {item[2]:<6} | {item[3]:<6} | {item[4]:<10}")
```

### 5. Pengujian pada Data Acak (Belum Terurut)
Menjalankan Binary Search pada data yang sengaja diacak untuk melihat kegagalan algoritma.

```python
import random

data_acak = data_sorted.copy()
random.shuffle(data_acak)

idx_acak, count_acak, _ = binary_search_trace(data_acak, 45)
print("Data Acak:", data_acak)
print(f"Hasil pencarian 45 pada data acak -> Indeks: {idx_acak}")
```

### 6. Analisis Hasil Pengujian Binary Search
* **Mengapa Binary Search Gagal pada Data Acak?**
  Binary Search mengasumsikan data sudah terurut sehingga dapat dengan yakin mengeliminasi separuh bagian array (kiri atau kanan) jika `target` tidak sama dengan `arr[mid]`. Pada data acak, pengabaian separuh bagian list akan membuat data yang dicari berisiko terbuang secara salah, sehingga menghasilkan output `-1` (tidak ditemukan) padahal nilainya ada.
* **Perbandingan Pemeriksaan dengan Linear Search:**
  Untuk $n = 20$, Linear Search pada kasus terburuk membutuhkan hingga **20 kali** pemeriksaan. Sementara itu, Binary Search pada kasus terburuk hanya membutuhkan maksimal $\lceil \log_2(20) \rceil = \mathbf{5\text{ kali}}$ pemeriksaan.

## 2.23 Mini Studi Kasus Terintegrasi
Menggabungkan seluruh operasi array, Linear Search, dan Binary Search pada sebuah studi kasus data nilai mahasiswa.

### 1 - 7. Eksekusi Program Mini Studi Kasus

```python
# Data kasus awal
nilai = [65, 72, 88, 75, 90, 72, 81, 77, 95, 68]
print("Data Awal:", nilai)

# 1. Tampilkan seluruh data beserta indeksnya menggunakan enumerate()
print("\n--- 1. Data dan Indeks ---")
for idx, val in enumerate(nilai):
    print(f"Indeks {idx}: {val}")

# 2. Cari nilai 90 menggunakan Linear Search
idx_90, _ = linear_search_trace(nilai, 90)
print(f"\n--- 2. Pencarian Nilai 90 ---")
print(f"Nilai 90 ditemukan pada indeks: {idx_90}")

# 3. Cari semua kemunculan nilai 72
kemunculan_72 = [idx for idx, val in enumerate(nilai) if val == 72]
print(f"\n--- 3. Kemunculan Nilai 72 ---")
print(f"Nilai 72 muncul pada indeks: {kemunculan_72}")

# 4. Ubah nilai pada indeks 0 menjadi 70
nilai[0] = 70
print(f"\n--- 4. Perubahan Indeks 0 menjadi 70 ---")
print("Data terbaru:", nilai)

# 5. Tambahkan nilai 84 di akhir dan sisipkan 76 pada indeks 3
nilai.append(84)
nilai.insert(3, 76)
print(f"\n--- 5. Tambah (append 84) & Sisip (insert 76 di idx 3) ---")
print("Data terbaru:", nilai)

# 6. Urutkan salinan data menggunakan sorted() tanpa mengubah data asli
nilai_sorted = sorted(nilai)
print(f"\n--- 6. Pengurutan Data ---")
print("Data Terurut (Salinan) :", nilai_sorted)
print("Data Asli (Tidka Berubah):", nilai)

# 7. Cari nilai 88 pada data terurut menggunakan Binary Search
idx_88_bs, _, _ = binary_search_trace(nilai_sorted, 88)
print(f"\n--- 7. Binary Search Nilai 88 pada Data Terurut ---")
print(f"Nilai 88 pada data terurut ditemukan pada indeks: {idx_88_bs}")
```

### 8. Ringkasan Kompleksitas Utama

| Operasi | Kompleksitas Waktu | Keterangan |
| :--- | :--- | :--- |
| **Akses Indeks** | $O(1)$ | Mengakses langsung lokasi memori tanpa pergeseran. |
| **Traversal** | $O(n)$ | Mengunjungi seluruh elemen sebanyak $n$ kali. |
| **Insert di Tengah** | $O(n)$ | Membutuhkan pergeseran elemen setelah posisi sisip. |
| **Linear Search** | $O(n)$ | Memeriksa data satu per satu dari awal sampai akhir. |
| **Binary Search** | $O(\log n)$ | Membagi ruang pencarian menjadi setengah pada tiap iterasi. |

### 9. Keputusan Algoritma: Pencarian Sekali pada Data Belum Terurut
Pilihan terbaik adalah **Linear Search**.

**Alasan:** Linear Search pada data acak membutuhkan waktu $O(n)$. Jika kita memilih Binary Search, kita diwajibkan melakukan sorting terlebih dahulu yang memakan waktu $O(n \log n)$, lalu melakukan pencarian $O(\log n)$, sehingga total kompleksitasnya menjadi $O(n \log n)$. Karena total $O(n) < O(n \log n)$, maka Linear Search jauh lebih efisien untuk pencarian tunggal.

### 10. Keputusan Algoritma: Pencarian Ribuan Kali pada Data Terurut
Pilihan terbaik adalah **Binary Search**.

**Alasan:** Karena data sudah dalam kondisi terurut sejak awal, kita tidak perlu mengeluarkan biaya untuk pengurutan data. Menjalankan Binary Search sebanyak $k$ kali membutuhkan waktu total $O(k \log n)$, sedangkan Linear Search membutuhkan waktu $O(k \cdot n)$. Untuk ribuan pencarian ($k$ bernilai besar), $O(k \log n)$ akan jauh lebih cepat secara signifikan dibanding Linear Search.
