**3.1 Membangun Linked List dari Nol**

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def insert_beginning(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def append(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            return
        current = self.head
        while current.next is not None:
            current = current.next
        current.next = new_node

    def display(self):
        current = self.head
        while current is not None:
            print(current.data, end=' -> ')
            current = current.next
        print('None')

# Uji Program
linked = SinglyLinkedList()
linked.append(10)
linked.append(20)
linked.append(30)
linked.insert_beginning(5)
linked.display()
```

**3.2 Operasi Insert dan Search**

```python
class Node:

  def __init__(self, data):
    self.data = data
    self.next = None


class SinglyLinkedList:

  def __init__(self):
    self.head = None

  # 2. Append (Tambah data di akhir)
  def append(self, data):
    new_node = Node(data)
    if self.head is None:
      self.head = new_node
      return
    current = self.head
    while current.next is not None:
      current = current.next
    current.next = new_node

  # 3. Insert Beginning (Tambah data di awal)
  def insert_beginning(self, data):
    new_node = Node(data)
    new_node.next = self.head
    self.head = new_node

  # 4. Insert After (Sisip data setelah nilai tertentu)
  def insert_after(self, target, data):
    current = self.head
    while current is not None:
      if current.data == target:
        new_node = Node(data)
        new_node.next = current.next
        current.next = new_node
        return True
      current = current.next
    return False

  # 5. Display
  def display(self):
    current = self.head
    while current is not None:
      print(current.data, end=" -> ")
      current = current.next
    print("None")

  # 6. Search
  def search(self, target):
    current = self.head
    pos = 0
    while current is not None:
      if current.data == target:
        return True, pos
      current = current.next
      pos += 1
    return False, -1


# --- Eksekusi Praktik 3.2 ---
sll = SinglyLinkedList()  # 1. Buat list kosong

# 2. Append 15, 25, 35, 45
sll.append(15)
sll.append(25)
sll.append(35)
sll.append(45)

# 3. Tambahkan 5 di awal
sll.insert_beginning(5)

# 4. Sisipkan 30 setelah nilai 25
sll.insert_after(25, 30)

# 5. Tampilkan linked list
print("Isi Linked List:")
sll.display()

# 6. Cari nilai 30 dan 99
status_30, pos_30 = sll.search(30)
status_99, pos_99 = sll.search(99)

print(
    f"Hasil Pencarian 30: Ditemukan ({status_30}) pada indeks/posisi ke-{pos_30}"
)
print(
    f"Hasil Pencarian 99: Ditemukan ({status_99}) pada indeks/posisi ke-{pos_99}"
)
```

**3.3 Delete dan Edge Case**

```python
# Tambahkan method delete dan count pada class
class SinglyLinkedListLengkap(SinglyLinkedList):

  def count(self):
    cnt = 0
    current = self.head
    while current:
      cnt += 1
      current = current.next
    return cnt

  def delete(self, key):
    current = self.head
    previous = None

    # Edge Case 1: List kosong
    if current is None:
      print(f"[Delete {key}] Gagal: List kosong.")
      return

    # Edge Case 2: Node yang dihapus adalah head
    if current.data == key:
      self.head = current.next
      current = None
      return

    # Menelusuri node yang akan dihapus
    while current is not None and current.data != key:
      previous = current
      current = current.next

    # Edge Case 3: Nilai tidak ditemukan
    if current is None:
      print(f"[Delete {key}] Gagal: Nilai {key} tidak ditemukan.")
      return

    # Menghapus node tengah/akhir
    previous.next = current.next
    current = None


# --- Eksekusi Pengujian Praktik 3.3 ---

# 1. Uji delete pada list kosong
print("=== 1. Delete pada List Kosong ===")
sll_demo = SinglyLinkedListLengkap()
sll_demo.delete(10)
sll_demo.display()
print("Jumlah node:", sll_demo.count())

# 2. List 1 node lalu hapus node tersebut
print("\n=== 2. Delete pada List 1 Node ===")
sll_demo.append(100)
print("Sebelum hapus:")
sll_demo.display()
sll_demo.delete(100)
print("Setelah hapus:")
sll_demo.display()
print("Jumlah node:", sll_demo.count())

# 3. List 5 node: hapus head, tengah, dan akhir
print("\n=== 3. Delete Head, Tengah, dan Akhir ===")
for val in [10, 20, 30, 40, 50]:
  sll_demo.append(val)

print("Awal:")
sll_demo.display()

sll_demo.delete(10)  # Hapus head
print("Setelah hapus Head (10):")
sll_demo.display()
print("Jumlah node:", sll_demo.count())

sll_demo.delete(30)  # Hapus tengah
print("Setelah hapus Tengah (30):")
sll_demo.display()
print("Jumlah node:", sll_demo.count())

sll_demo.delete(50)  # Hapus akhir
print("Setelah hapus Akhir (50):")
sll_demo.display()
print("Jumlah node:", sll_demo.count())

# 4. Hapus nilai yang tidak ada
print("\n=== 4. Delete Nilai yang Tidak Ada ===")
sll_demo.delete(999)
print("Isi list akhir:")
sll_demo.display()
print("Jumlah node:", sll_demo.count())
```
