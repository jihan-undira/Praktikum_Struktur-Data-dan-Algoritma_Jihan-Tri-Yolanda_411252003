**1: Membuat Class Queue Berbasis collections.deque**

```python
from collections import deque

class QueueBase:
    def __init__(self):
        self._items = deque()

    def enqueue(self, item):
        self._items.append(item)

    def dequeue(self):
        if self.is_empty():
            print("Peringatan: Queue kosong!")
            return None
        return self._items.popleft()

    def peek(self):
        if self.is_empty():
            print("Peringatan: Queue kosong!")
            return None
        return self._items[0]

    def size(self):
        return len(self._items)

    def is_empty(self):
        return len(self._items) == 0
```

**9: Bukti Urutan Keluar Mengikuti FIFO**

Prinsip FIFO (First-In, First-Out) menegaskan bahwa data yang pertama kali dimasukkan akan menjadi data yang pertama kali dikeluarkan.

Urutan Kedatangan Awal: Data dimasukkan berurutan: 10, 20, 30, 40, 50.

Pengeluaran Pertama: Saat dequeue() dipanggil dua kali, elemen yang keluar adalah 10 lalu 20. Hal ini membuktikan 10 (yang paling pertama masuk) keluar lebih awal dibanding elemen yang masuk setelahnya.

Penambahan Data Baru: 60 dan 70 dimasukkan kemudian, sehingga berada di posisi paling belakang (rear).

Pengeluaran Akhir: Saat di-dequeue sampai kosong, urutan keluarnya adalah 30, 40, 50, 60, 70. 30 keluar lebih dahulu dibanding 60 dan 70 karena 30 berada di antrean lebih lama.

Setiap operasi pengeluaran konsisten mengambil elemen di posisi front (posisi data paling awal yang belum diproses), yang membuktikan perilaku FIFO.

# Praktik Mandiri 5.2 — Circular Queue

**1 s.d. 6. Implementasi Kode & Eksekusi Langkah demi Langkah**

```python
class CircularQueue:
    def __init__(self, capacity):
        self.capacity = capacity
        self.items = [None] * capacity
        self.front = 0
        self.rear = -1
        self.count = 0

    def is_full(self):
        return self.count == self.capacity

    def is_empty(self):
        return self.count == 0

    def enqueue(self, item):
        print(f"\n[Operasi]: Enqueue('{item}')")
        if self.is_full():
            print(f"-> GAGAL: Queue Penuh! Tidak dapat menambahkan '{item}'.")
            self.display_state()
            return False

        self.rear = (self.rear + 1) % self.capacity
        self.items[self.rear] = item
        self.count += 1
        self.display_state()
        return True

    def dequeue(self):
        print("\n[Operasi]: Dequeue()")
        if self.is_empty():
            print("-> GAGAL: Queue Kosong!")
            self.display_state()
            return None

        removed = self.items[self.front]
        self.items[self.front] = None  # Reset slot
        self.front = (self.front + 1) % self.capacity
        self.count -= 1
        print(f"-> Berhasil mengeluarkan: '{removed}'")
        self.display_state()
        return removed

    def display_state(self):
        print(f"   Items (Array Fisik) : {self.items}")
        print(f"   Front: {self.front} | Rear: {self.rear} | Count: {self.count}")

# --- EKSEKUSI SKENARIO (Langkah 1 - 5) ---

# 1. Buat CircularQueue dengan kapasitas 4
cq = CircularQueue(4)
print("=== CircularQueue Berkapasitas 4 Dibuat ===")

# 2. Enqueue A, B, C, D dan cek kondisi penuh
for item in ['A', 'B', 'C', 'D']:
    cq.enqueue(item)

print(f"\nApakah Queue Penuh? -> {cq.is_full()}")

# 3. Coba enqueue E ketika penuh
cq.enqueue('E')

# 4. Dequeue dua elemen
cq.dequeue()
cq.dequeue()

# 5. Enqueue E dan F (Rear mengalami wrap-around)
cq.enqueue('E')
cq.enqueue('F')
```

**7. Keadaan Array Fisik vs Urutan Queue Logis**

Setelah Enqueue A, B, C, D (Kondisi Penuh Awal):
* Array Fisik: ['A', 'B', 'C', 'D']
* Pointer: front = 0, rear = 3, count 4
* Urutan Logis (Front $\rightarrow$ Rear): A $\rightarrow$ B $\rightarrow$ C $\rightarrow$ D

Setelah Dequeue 2 Elemen ('A' dan 'B'):
* Array Fisik: [None, None, 'C', 'D']
* Pointer: front = 2, rear = 3, count = 2
* Urutan Logis (Front $\rightarrow$ Rear): C $\rightarrow$ D

Setelah Enqueue 'E' dan 'F' (Wrap-Around):
* Array Fisik: ['E', 'F', 'C', 'D']
* Pointer: front = 2, rear = 1, count = 4
* Urutan Logis (Front $\rightarrow$ Rear): C $\rightarrow$ D $\rightarrow$ E $\rightarrow$

**8. Fungsi Operasi Modulo pada Implementasi**

Operasi modulo (% capacity) berfungsi untuk memutar kembali indeks ke awal array (indeks 0) saat indeks melampaui batas kapasitas maksimum.
* Formula Rear Next: (rear + 1) % capacity
* Formula Front Next: (front + 1) % capacity

Contoh Kasus Pada Simulasi Di Atas:
1.   Sebelum memasukkan E, pointer rear berada di indeks terakhir (3).
2.   Perhitungan indeks berikutnya: (3 + 1) % 4 = 4 % 4 = 0.
3. Indeks berputar (wrap-around) kembali ke 0 tanpa menyebabkan error Index out of bounds.

Tanpa modulo, kita harus menggeser seluruh elemen tersisa ke kiri setiap kali dequeue() dilakukan ($\mathcal{0}(n)$). Modulo memungkinkan penggunaan kembali (reuse) slot kosong secara efisien tanpa pergeseran data.

**9. Kompleksitas Operasi**

* Enqueue (enqueue):
  * Kompleksitas Waktu (Time Complexity): $\mathcal{O}(1)$
  * Kompleksitas Ruang (Space Complexity): $\mathcal{O}(1)$
* Dequeue (dequeue):
  * Kompleksitas Waktu (Time Complexity): $\mathcal{O}(1)$
  * Kompleksitas Ruang (Space Complexity): $\mathcal{O}(1)$
  
Penjelasan: Baik enqueue maupun dequeue hanya melakukan kalkulasi aritmatika sederhana (operasi modulo), penunjukan indeks array, serta pembaharuan variabel pointer (front, rear, count). Tidak ada perulangan (loop) maupun pemindahan elemen array.

# Praktik Mandiri 5.3 — Deque dan Palindrome

**1: Implementasi Kembali Fungsi**

```python
from collections import deque
import re

def is_palindrome(text: str, debug: bool = False) -> bool:
    # Langkah normalisasi (dijelaskan pada Nomor 3)
    normalized = re.sub(r'[^a-zA-Z0-9]', '', text).lower()
    char_deque = deque(normalized)

    while len(char_deque) > 1:
        if char_deque.popleft() != char_deque.pop():
            return False

    return True
```

**2: Pengujian Kata Tunggal**

```python
words = ["level", "katak", "radar", "python", "algoritma"]

print("=== UJI KATA TUNGGAL ===")
for word in words:
    result = is_palindrome(word)
    print(f"Kata: '{word:<10}' -> Palindrome: {result}")
```

**3: Pengujian Kalimat dengan Normalisasi**

```python
phrases = [
    "Kasur ini rusak",
    "Madam, I'm Adam",
    "A man, a plan, a canal: Panama",
    "Struktur Data"
]

print("\n=== UJI KALIMAT DENGAN NORMALISASI ===")
for phrase in phrases:
    normalized = re.sub(r'[^a-zA-Z0-9]', '', phrase).lower()
    result = is_palindrome(phrase)
    print(f"Kalimat   : '{phrase}'")
    print(f"Normalisasi: '{normalized}' -> Palindrome: {result}\n")
```

**4 & 5: Fungsi Modifikasi dengan Tampilan Iterasi Kiri-Kanan & Tabel Dry Run**

```python
def is_palindrome_debug(text: str) -> bool:
    normalized = re.sub(r'[^a-zA-Z0-9]', '', text).lower()
    char_deque = deque(normalized)

    print(f"\n--- Pengujian Debug: '{text}' ---")
    print(f"Hasil Normalisasi: '{normalized}'")

    iteration = 1
    while len(char_deque) > 1:
        left = char_deque.popleft()
        right = char_deque.pop()
        match_status = "SAMA" if left == right else "BEDA"

        print(f"Iterasi {iteration}: Kiri='{left}' | Kanan='{right}' -> {match_status}")

        if left != right:
            return False
        iteration += 1

    return True

# Menjalankan fungsi debug
is_palindrome_debug("Kasur ini rusak")
```

**6: Mengapa Deque Lebih Natural daripada Queue Satu Arah?**

1. Akses Akses Dua Ujung: Permasalahan palindrom menuntut ekstraksi karakter dari ujung depan dan ujung belakang secara bersamaan.
2. Keterbatasan Queue Satu Arah: Queue standar hanya mengizinkan pengeluaran elemen dari ujung depan (dequeue). Untuk mengambil elemen paling belakang pada Queue biasa, seluruh elemen di depannya harus dikeluarkan terlebih dahulu (atau memerlukan struktur data tambahan seperti Stack).
3. Efisiensi: Deque menyediakan operasi popleft() dan pop() yang keduanya berjalan pada kompleksitas waktu konstan $\mathcal{O}(1)$, menjadikannya pilihan struktur data paling alami.

**7: Kompleksitas Waktu dan Ruang**

* Kompleksitas Waktu (Time Complexity): $\mathcal{O}(n)$
  * Normalisasi string memproses setiap karakter sepanjang $n$ dalam waktu $\mathcal{O}(n)$.
  * Menyimpan $n$ karakter ke Deque membutuhkan waktu $\mathcal{O}(n)$.
  * Perbandingan karakter dalam perulangan berjalan sebanyak maksimal $\lfloor n/2 \rfloor$ kali, di mana operasi popleft() dan pop() bernilai $\mathcal{O}(1)$.
  * Total kompleksitas waktu adalah linier $\mathcal{O}(n)$.
* Kompleksitas Ruang (Space Complexity): $\mathcal{O}(n)$
  * Membutuhkan memori tambahan untuk menyimpan string hasil normalisasi serta elemen-elemen karakter di dalam Deque sebanyak $n$ elemen.
