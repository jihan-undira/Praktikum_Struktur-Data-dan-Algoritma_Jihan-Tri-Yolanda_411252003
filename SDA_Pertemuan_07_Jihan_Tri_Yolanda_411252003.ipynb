# Modul 7: Praktik Mandiri Algoritma Pengurutan (Sorting)

```python
# ==============================================================================
# 7.25. PRAKTIK MANDIRI 7.1 — BUBBLE SORT DESCENDING
# ==============================================================================

def bubble_sort_desc(data):
    arr = data.copy()
    n = len(arr)
    print(f"Data awal: {arr}\n")

    for i in range(n - 1):
        swapped = False
        for j in range(0, n - i - 1):
            # Perbandingan diubah menjadi < agar elemen lebih kecil digeser ke kanan
            if arr[j] < arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        print(f"Pass {i + 1}: {arr}")
        if not swapped:
            break
    return arr

print("=== 7.25. BUBBLE SORT DESCENDING ===")
data_7_1 = [17, 4, 25, 9, 13, 2]
hasil_7_1 = bubble_sort_desc(data_7_1)
print(f"Target Output: {hasil_7_1}\n")
```

## Analisis 7.25 — Bubble Sort Descending

1. **Kondisi Perbandingan yang Diubah:**
   * Pada pengurutan *ascending* (terkecil ke terbesar), kondisi pertukaran adalah `arr[j] > arr[j + 1]`.
   * Untuk *descending* (terbesar ke terkecil), kondisi diubah menjadi `arr[j] < arr[j + 1]`. Jika elemen kiri lebih kecil dari elemen kanan, kedua elemen ditukar posisi agar nilai yang lebih besar bergeser ke kiri.

2. **Kompleksitas Kasus (Best & Worst Case):**
   * **Best Case — $\mathcal{O}(n)$:** Terjadi jika data awal sudah terurut *descending* sempurna (misal: `[25, 17, 13, 9, 4, 2]`). Dengan flag pertukaran (*swapped*), algoritma berhenti hanya dalam 1 *pass* ($\mathcal{O}(n)$ perbandingan, 0 pertukaran).
   * **Worst Case — $\mathcal{O}(n^2)$:** Terjadi jika data awal terurut secara *ascending* terbalik (misal: `[2, 4, 9, 13, 17, 25]`). Membutuhkan $n(n-1)/2$ kali perbandingan dan pertukaran penuh.

```python
# ==============================================================================
# 7.26. PRAKTIK MANDIRI 7.2 — SELECTION SORT DENGAN STATISTIK
# ==============================================================================

def selection_sort_stats(data):
    arr = data.copy()
    n = len(arr)
    comparisons = 0
    swaps = 0

    for i in range(n - 1):
        min_idx = i
        for j in range(i + 1, n):
            comparisons += 1
            if arr[j] < arr[min_idx]:
                min_idx = j

        if min_idx != i:
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
            swaps += 1

    return arr, comparisons, swaps

print("=== 7.26. SELECTION SORT DENGAN STATISTIK ===")
uji_urut = [1, 2, 3, 4, 5, 6]
uji_acak = [4, 1, 6, 2, 5, 3]
uji_terbalik = [6, 5, 4, 3, 2, 1]

for nama, d in [("Terurut", uji_urut), ("Acak", uji_acak), ("Terbalik", uji_terbalik)]:
    res, comp, swp = selection_sort_stats(d)
    print(f"Data {nama:<8}: Hasil={res} | Comparisons={comp} | Swaps={swp}")
print()
```

## Analisis 7.26 — Selection Sort dengan Statistik

* **Apakah jumlah perbandingan berubah?**
  **Tidak.** Pada ketiga kondisi data (terurut, acak, maupun terbalik), jumlah perbandingan yang dihasilkan **tetap sama persis yaitu 15 kali** ($\frac{6 \times 5}{2} = 15$).

* **Hubungan dengan Kompleksitas $\mathcal{O}(n^2)$:**
  Selection Sort bekerja dengan cara mencari nilai minimum dari sisa *sub-array* yang belum terurut pada setiap iterasi. Loop luar berjalan $n-1$ kali dan loop dalam berjalan dari $i+1$ hingga $n-1$. Perbandingan dilakukan secara mutlak tanpa peduli susunan data awal. Rumus total perbandingannya adalah:
  $$\text{Total Perbandingan} = \sum_{i=1}^{n-1} i = \frac{n(n-1)}{2} = \frac{n^2 - n}{2}$$
  Karena suku dominannya adalah $n^2$, kompleksitas waktu untuk *best*, *average*, maupun *worst case* dari perbandingan Selection Sort adalah konstan di $\mathcal{O}(n^2)$.

```python
# ==============================================================================
# 7.27. PRAKTIK MANDIRI 7.3 — INSERTION SORT DAN DATA HAMPIR TERURUT
# ==============================================================================

def insertion_sort_shifts(data):
    arr = data.copy()
    shifts = 0
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]  # Pergeseran elemen
            shifts += 1
            j -= 1
        arr[j + 1] = key
    return arr, shifts

print("=== 7.27. INSERTION SORT DAN DATA HAMPIR TERURUT ===")
hampir = [1, 2, 3, 5, 4, 6, 7, 8]
acak = [7, 2, 8, 1, 5, 3, 6, 4]

_, shift_hampir = insertion_sort_shifts(hampir)
_, shift_acak = insertion_sort_shifts(acak)

print(f"Jumlah pergeseran data hampir terurut (n=8) : {shift_hampir}")
print(f"Jumlah pergeseran data acak (n=8)          : {shift_acak}\n")

# Uji dengan 20 elemen
import random
random.seed(42)

hampir_20 = list(range(1, 21))
hampir_20[8], hampir_20[9] = hampir_20[9], hampir_20[8] # Tukar 1 pasangan elemen
acak_20 = random.sample(range(1, 100), 20)

_, shift_hampir_20 = insertion_sort_shifts(hampir_20)
_, shift_acak_20 = insertion_sort_shifts(acak_20)

print("--- Uji 20 Elemen ---")
print(f"Jumlah pergeseran data hampir terurut (n=20): {shift_hampir_20}")
print(f"Jumlah pergeseran data acak (n=20)         : {shift_acak_20}\n")
```

## Analisis 7.27 — Insertion Sort dan Data Hampir Terurut

1. **Perbandingan Jumlah Pergeseran:**
   * Pada data $n=8$: Data hampir terurut hanya memerlukan **1 pergeseran**, sedangkan data acak memerlukan **13 pergeseran**.
   * Pada data $n=20$: Data hampir terurut hanya memerlukan **1 pergeseran**, sedangkan data acak memerlukan hingga **93 pergeseran**.

2. **Mengapa Insertion Sort Sangat Menarik untuk Data Hampir Terurut?**
   * Insertion Sort bekerja secara adaptif. Pada loop dalam (`while j >= 0 and arr[j] > key`), jika elemen di sebelah kiri sudah lebih kecil dari `key`, loop akan langsung berhenti (*break early*).
   * Untuk data hampir terurut, mayoritas elemen sudah berada di posisi yang tepat sehingga pergeseran elemen bernilai mendekati nol. Hal ini membuat waktu eksekusi Insertion Sort mendekati **linier $\mathcal{O}(n)$**, jauh lebih cepat dibanding algoritma $\mathcal{O}(n \log n)$ yang memiliki *overhead* rekursi tinggi.

```python
# ==============================================================================
# 7.28. PRAKTIK MANDIRI 7.4 — MERGE SORT DENGAN TRACE
# ==============================================================================

def merge_sort_trace(arr, level=0, label="Root"):
    indent = "  " * level
    print(f"{indent}[Level {level} - {label}] Split: {arr}")

    # Base case: elemen <= 1
    if len(arr) <= 1:
        print(f"{indent}  -> Reached Base Case: {arr}")
        return arr

    mid = len(arr) // 2
    left_half = merge_sort_trace(arr[:mid], level + 1, "Left")
    right_half = merge_sort_trace(arr[mid:], level + 1, "Right")

    # Process merge
    merged = []
    i = j = 0
    while i < len(left_half) and j < len(right_half):
        if left_half[i] <= right_half[j]:
            merged.append(left_half[i])
            i += 1
        else:
            merged.append(right_half[j])
            j += 1
    merged.extend(left_half[i:])
    merged.extend(right_half[j:])

    print(f"{indent}  -> Merge Result: {merged}")
    return merged

print("=== 7.28. MERGE SORT TRACE ===")
data_7_4 = [21, 4, 16, 8, 15, 2, 10, 7]
merge_sort_trace(data_7_4)
print()
```

## Analisis 7.28 — Merge Sort dengan Trace

1. **Base Case yang Tercapai:**
   Base case tercapai ketika panjang *sub-array* $\le 1$. Total ada **8 base case** individual yang tercapai (setiap elemen tunggal):  
   `[21]`, `[4]`, `[16]`, `[8]`, `[15]`, `[2]`, `[10]`, `[7]`.

2. **Urutan Hasil Merge dari Level Terdalam hingga Final:**
   * Level 3 (Pasangan Elemen):
     1. `merge([21], [4])` $\rightarrow$ `[4, 21]`
     2. `merge([16], [8])` $\rightarrow$ `[8, 16]`
     3. `merge([15], [2])` $\rightarrow$ `[2, 15]`
     4. `merge([10], [7])` $\rightarrow$ `[7, 10]`
   * Level 2 (Empat Elemen):
     5. `merge([4, 21], [8, 16])` $\rightarrow$ `[4, 8, 16, 21]`
     6. `merge([2, 15], [7, 10])` $\rightarrow$ `[2, 7, 10, 15]`
   * Level 1 (Final Merge - Delapan Elemen):
     7. `merge([4, 8, 16, 21], [2, 7, 10, 15])` $\rightarrow$ `[2, 4, 7, 8, 10, 15, 16, 21]`

3. **Mengapa Banyak Level Pembagian Mendekati $\log_2 n$?**
   * Pada setiap langkah pembagian (*divide*), array berukuran $n$ dibagi menjadi dua bagian yang berukuran sama ($n/2$).
   * Proses pembagian berhenti ketika ukuran array menjadi 1. Persamaan matematikanya adalah:
     $$\frac{n}{2^k} = 1 \implies 2^k = n \implies k = \log_2 n$$
   * Di mana $k$ adalah kedalaman/banyaknya level pohon rekursi. Untuk $n=8$, levelnya adalah $\log_2 8 = 3$ tingkat pembagian.

```python
# ==============================================================================
# 7.29. PRAKTIK MANDIRI 7.5 — EKSPERIMEN EMPIRIS (PERBAIKAN)
# ==============================================================================

import timeit
import random

def bubble_sort(arr):
    a = arr.copy()
    n = len(a)
    for i in range(n):
        for j in range(0, n - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
    return a

def insertion_sort(arr):
    a = arr.copy()
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = key
    return a

def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    res = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            res.append(left[i])
            i += 1
        else:
            res.append(right[j])
            j += 1
    res.extend(left[i:])
    res.extend(right[j:])
    return res

sizes = [100, 500, 1000]
results = []

for n in sizes:
    random.seed(42)
    base_data = [random.randint(1, 10000) for _ in range(n)]

    t_bubble = timeit.timeit(lambda: bubble_sort(base_data), number=5) / 5
    t_insert = timeit.timeit(lambda: insertion_sort(base_data), number=5) / 5
    t_merge  = timeit.timeit(lambda: merge_sort(base_data), number=5) / 5

    results.append({
        "n": n,
        "Bubble Sort (s)": round(t_bubble, 6),
        "Insertion Sort (s)": round(t_insert, 6),
        "Merge Sort (s)": round(t_merge, 6)
    })

print("=== 7.29. TABEL HASIL EKSPERIMEN EMPIRIS (WAKTU EKSEKUSI DALAM DETIK) ===")
print(f"{'n':<6} | {'Bubble Sort (s)':<18} | {'Insertion Sort (s)':<18} | {'Merge Sort (s)':<18}")
print("-" * 68)
for row in results:
    print(f"{row['n']:<6} | {row['Bubble Sort (s)']:<18} | {row['Insertion Sort (s)']:<18} | {row['Merge Sort (s)']:<18}")
```

## Analisis 7.29 — Eksperimen Empiris

1. **Kesesuaian dengan Teori $\mathcal{O}(n^2)$ vs $\mathcal{O}(n \log n)$:**
   * **Sesuai.** Saat $n$ naik 10 kali lipat (dari 100 ke 1.000):
     * Algoritma $\mathcal{O}(n^2)$ (Bubble Sort & Insertion Sort) mengalami kenaikan waktu eksekusi secara kuadratik (mendekati $10^2 = 100$ kali lipat).
     * Algoritma $\mathcal{O}(n \log n)$ (Merge Sort) mengalami kenaikan waktu yang jauh lebih landai dan linier. Pada $n=1.000$, Merge Sort terbukti jauh lebih cepat dibanding Bubble/Insertion Sort.

2. **Mengapa Runtime Nyata Tidak Selalu Mengikuti Rasio Teori Secara Persis?**
   * **Overhead Bahasa & Rekursi:** Merge Sort membutuhkan alokasi memori baru (*slicing list*) dan panggilan fungsi rekursif berulang (*function call overhead*), sehingga pada $n$ yang sangat kecil ($n=100$), eksekusinya bisa sedikit lebih lambat atau setara dengan Insertion Sort yang sederhana.
   * **Cache & Memory Management:** Penyimpanan memori dan optimasi interpreter Python mempengaruhi pembacaan data di CPU cache.
   * **Faktor Konstanta Interupsi OS:** Kompleksitas Notasi Big-O mengabaikan konstanta $c$ dan suku tingkat rendah ($c \cdot f(n)$). Selain itu, *CPU context switching* dari sistem operasi saat menjalankan `timeit` juga menciptakan sedikit variasi waktu komputasi nyata.
