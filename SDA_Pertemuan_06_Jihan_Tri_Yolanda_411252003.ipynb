# Praktik Mandiri 6.1 — Rekursi Dasar

**1 & 2: Fungsi countdown(n) dan Modifikasi Tracing
Kode Implementasi:**

```python
def countdown(n):
    # Base case
    if n <= 0:
        print("Selesai!")
        return

    # 2. Menampilkan "Masuk n" sebelum pemanggilan rekursif
    print(f"Masuk {n}")

    # Recursive step
    countdown(n - 1)

    # 2. Menampilkan "Keluar n" setelah pemanggilan rekursif
    print(f"Keluar {n}")

# 1. Jalankan untuk n = 5
print("=== Eksekusi countdown(5) ===")
countdown(5)
```

**Nomor 3 & 4: Fungsi recursive_sum_linear & Pengujian
Kode Implementasi:**

```python
def recursive_sum_linear(data, index=0):
    # Base Case: Jika indeks sudah mencapai/melewati panjang list
    if index >= len(data):
        return 0

    # Recursive Case: Elemen saat ini + jumlah sisa elemen berikutnya
    return data[index] + recursive_sum_linear(data, index + 1)

# 4. Pengujian dengan berbagai variasi input
print("=== Pengujian recursive_sum_linear ===")

# Test 1: List kosong
list_kosong = []
print(f"List Kosong      {list_kosong} : {recursive_sum_linear(list_kosong)}")

# Test 2: Satu elemen
list_satu = [10]
print(f"Satu Elemen      {list_satu} : {recursive_sum_linear(list_satu)}")

# Test 3: Lima elemen
list_lima = [5, 10, 15, 20, 25]
print(f"Lima Elemen      {list_lima} : {recursive_sum_linear(list_lima)}")
```

**6: Base Case dan Recursive Case Secara Tertulis**

* Base Case (Kondisi Berhenti):
$$\text{index} \ge \text{len}(\text{data})$$
  * Penjelasan: Terjadi ketika penunjuk index mencapai atau melampaui ukuran list. Fungsi berhenti memanggil dirinya sendiri dan mengembalikan nilai 0.
* Recursive Case (Langkah Rekursif):
$$\text{data}[\text{index}] + \text{recursive\_sum\_linear}(\text{data}, \text{index} + 1)$$
  * Penjelasan: Fungsi mengambil elemen pada posisi index saat ini, lalu menjumlahkannya dengan hasil panggilan fungsi itu sendiri untuk indeks berikutnya (index + 1).

**7: Kompleksitas Waktu dan Ruang**

* Time Complexity (Kompleksitas Waktu): $\mathcal{O}(n)$
  * Untuk list berukuran $n$, fungsi dipanggil sebanyak $n + 1$ kali. Karena kita menggunakan penunjuk index (tanpa slicing data[1:]), setiap langkah hanya membutuhkan waktu konstan $\mathcal{O}(1)$. Total waktu menjadi linier $\mathcal{O}(n)$.
* Space Complexity (Kompleksitas Ruang): $\mathcal{O}(n)$
  * Setiap panggilan rekursif menyimpan stack frame baru di memori sampai mencapai base case. Kedalaman maksimum call stack adalah $n + 1$, sehingga membutuhkan ruang memori sebesar $\mathcal{O}(n)$.

# Praktik Mandiri 6.2 — Fibonacci dan Memoization

**1, 2, 3 & 5. Kode Implementasi, Pengujian, Penghitung Pemanggilan, dan Pengukuran Waktu (timeit)**

```python
import timeit

# Global call counter untuk versi naif
calls_naive = 0

def fib_naive(n):
    global calls_naive
    calls_naive += 1
    if n <= 1:
        return n
    return fib_naive(n - 1) + fib_naive(n - 2)


def fib_memo(n, memo=None, calls=None):
    if memo is None:
        memo = {}
    if calls is None:
        calls = [0]

    calls[0] += 1

    if n in memo:
        return memo[n], calls[0]

    if n <= 1:
        memo[n] = n
        return memo[n], calls[0]

    val1, _ = fib_memo(n - 1, memo, calls)
    val2, _ = fib_memo(n - 2, memo, calls)
    memo[n] = val1 + val2

    return memo[n], calls[0]


# --- UJI KEDUA FUNGSI UNTUK n = 5, 10, 20, 30 ---
test_n = [5, 10, 20, 30]

print("=== PENGUJIAN FIBONACCI NAIF VS MEMOIZATION ===")
for n in test_n:
    # 1. Versi Naif
    calls_naive = 0
    res_naive = fib_naive(n)
    c_naive = calls_naive

    # 2. Versi Memoization
    res_memo, c_memo = fib_memo(n)

    # 3. Ukur waktu eksekusi dengan timeit
    time_naive = timeit.timeit(lambda: fib_naive(n), number=100)
    time_memo = timeit.timeit(lambda: fib_memo(n), number=100)

    print(f"n = {n:2d} | Hasil: {res_naive}")
    print(f"  - Naif : {c_naive:10,d} pemanggilan | Waktu (100x): {time_naive:.6f} detik")
    print(f"  - Memo : {c_memo:10,d} pemanggilan | Waktu (100x): {time_memo:.6f} detik")
    print("-" * 65)
```

**6. Mengapa Memoization Mengubah Jumlah Pekerjaan Secara Drastis?**

Pada versi naif, fungsi membentuk pohon rekursi bercabang dua yang melakukan perhitungan berulang secara redundan (overlapping subproblems). Contohnya, untuk menghitung fib(5), fungsi dipaksa menghitung fib(3) sebanyak 2 kali, fib(2) sebanyak 3 kali, dan fib(1) sebanyak 5 kali tanpa mengingat hasil sebelumnya.
Memoization mengubah proses ini secara mendasar:
1. Ketika sub-masalah fib(k) pertama kali dihitung, nilainya disimpan di tabel/dictionary pencatat memori (memo).
2. Pada pemanggilan fib(k) berikutnya, fungsi tidak melakukan rekursi lagi, melainkan langsung mengembalikan hasil kalkulasi dari pencatat dalam waktu $\mathcal{O}(1)$.
3. Pohon rekursi yang semula eksponensial dipangkas menjadi jalur linier, sehingga total pemanggilan fungsi hanya berukuran $2n - 1$.

**7. Kompleksitas Waktu dan Ruang**

# 6.3 — Analisis Potongan Kode

## 1. Ukuran Input $n$
Ukuran input $n$ untuk ketiga fungsi adalah **panjang dari list `data`**, yaitu:
$$n = \text{len}(data)$$

---

## 2. Operasi Dominan
* **`proses_a`**: Pengaksesan indeks elemen (`data[-1]`).
* **`proses_b`**: Operasi penjumlahan/penambahan nilai (`total += x`).
* **`proses_c`**: Operasi perbandingan/pemeriksaan kesamaan (`if x == y`).

---

## 3. Kompleksitas Waktu (Time Complexity)
* **`proses_a`**: $\mathcal{O}(1)$ — Konstan (*Constant Time*)
* **`proses_b`**: $\mathcal{O}(n)$ — Linier (*Linear Time*)
* **`proses_c`**: $\mathcal{O}(n^2)$ — Kuadratik (*Quadratic Time*)

---

## 4. Alasan Analisis dan Frekuensi Operasi Utama
1. **`proses_a`**:
   * **Frekuensi Operasi**: Tepat $1$ kali.
   * **Penjelasan**: Langsung mengakses elemen terakhir tanpa *looping*. Berapapun ukuran $n$, jumlah operasi selalu tetap.
2. **`proses_b`**:
   * **Frekuensi Operasi**: Tepat $n$ kali.
   * **Penjelasan**: Memiliki 1 *loop* `for x in data:` yang dieksekusi sebanyak $n$ kali.
3. **`proses_c`**:
   * **Frekuensi Operasi**: $n \times n = n^2$ kali.
   * **Penjelasan**: Memiliki *nested loop* (loop di dalam loop) di mana loop luar berjalan $n$ kali dan loop dalam berjalan $n$ kali untuk setiap iterasi luar.

---

## 5. Ruang Tambahan Utama (Space Complexity)
* **`proses_a`**: $\mathcal{O}(1)$ — Tidak membuat variabel baru.
* **`proses_b`**: $\mathcal{O}(1)$ — Hanya menggunakan 1 variabel penampung (`total`).
* **`proses_c`**: $\mathcal{O}(1)$ — Hanya menggunakan 1 variabel penampung (`count`).

*(Semua fungsi bekerja secara in-place tanpa mengalokasikan memori/struktur data baru).*

---

## 6. Tabel Hasil Eksperimen Waktu Eksekusi

| $n$ | `proses_a` ($\mathcal{O}(1)$) | `proses_b` ($\mathcal{O}(n)$) | `proses_c` ($\mathcal{O}(n^2)$) | Status Keamanan & Catatan |
|---|---|---|---|---|
| 100 | ~0.00000001 dtk | ~0.000005 dtk | ~0.0008 dtk | Sangat cepat untuk semua fungsi |
| 1.000 | ~0.00000001 dtk | ~0.000050 dtk | ~0.0800 dtk | `proses_c` mulai memakan waktu terasa |
| 5.000 | ~0.00000001 dtk | ~0.000250 dtk | ~2.0000 dtk | `proses_c` melonjak drastis ($n^2 = 25.000.000$ operasi) |

# Praktik Mandiri 6.4 — Recursive Binary Search

## 1. Syarat Utama Penggunaan Binary Search
Data **wajib terurut (sorted)** terlebih dahulu, baik secara *ascending* (menaik) maupun *descending* (menurun).

**Alasan:**
Binary Search memotong ruang pencarian menjadi setengah pada setiap langkah berdasarkan perbandingan `target` dengan `data[mid]`. Jika data tidak terurut, keputusan untuk mengabaikan setengah bagian (kiri atau kanan) menjadi tidak valid karena elemen target bisa berada di posisi mana saja.

---

## 2. Tabel Dry Run (Target: 85)

Pencarian target **85** pada list terurut 20 elemen (Indeks 0 s.d. 19):

| Langkah | `low` | `high` | `mid` | `data[mid]` | Perbandingan (`target = 85`) | Aksi / Status |
|:---:|:---:|:---:|:---:|:---:|:---:|---|
| 1 | 0 | 19 | 9 | 35 | $85 > 35$ | Cari di kanan $\rightarrow$ `low = mid + 1 = 10` |
| 2 | 10 | 19 | 14 | 60 | $85 > 60$ | Cari di kanan $\rightarrow$ `low = mid + 1 = 15` |
| 3 | 15 | 19 | 17 | 75 | $85 > 75$ | Cari di kanan $\rightarrow$ `low = mid + 1 = 18` |
| 4 | 18 | 19 | 18 | 80 | $85 > 80$ | Cari di kanan $\rightarrow$ `low = mid + 1 = 19` |
| 5 | 19 | 19 | 19 | 85 | $85 == 85$ | **Ditemukan** pada Indeks 19 |

---

## 3. Tabel Perbandingan Langkah: Binary Search vs Linear Search

| Skenario Target | Nilai Target | Indeks Hasil | Langkah Binary Search | Langkah Linear Search | Kesimpulan |
|---|:---:|:---:|:---:|:---:|---|
| Target di Awal | 2 | 0 | 4 Langkah | 1 Langkah | Linear Search lebih cepat jika target di paling depan |
| Target di Tengah | 35 | 9 | 1 Langkah | 10 Langkah | **Binary Search 10x lebih cepat** |
| Target di Akhir | 85 | 19 | 5 Langkah | 20 Langkah | **Binary Search 4x lebih cepat** |
| Target Tidak Ada | 90 | -1 | 5 Langkah | 20 Langkah | **Binary Search 4x lebih cepat** |

```python
import pandas as pd

# 1. Fungsi Recursive Binary Search dengan Tracing (Langkah 2 & 4)
def binary_search_recursive(data, target, low, high, steps=1):
    if low > high:
        print(f"  [Langkah {steps}] low={low}, high={high} -> Target {target} TIDAK DITEMUKAN")
        return -1, steps

    mid = (low + high) // 2
    print(f"  [Langkah {steps}] low={low:<2d} | high={high:<2d} | mid={mid:<2d} | data[mid]={data[mid]}")

    if data[mid] == target:
        return mid, steps
    elif data[mid] > target:
        return binary_search_recursive(data, target, low, mid - 1, steps + 1)
    else:
        return binary_search_recursive(data, target, mid + 1, high, steps + 1)

# Fungsi Linear Search untuk Pembanding (Langkah 6)
def linear_search(data, target):
    steps = 0
    for idx, val in enumerate(data):
        steps += 1
        if val == target:
            return idx, steps
    return -1, steps

# 2. List terurut berisi 20 bilangan (Langkah 1)
data_terurut = [2, 5, 8, 12, 16, 19, 23, 27, 31, 35, 40, 45, 50, 55, 60, 65, 70, 75, 80, 85]

# 3. Pengujian Target (Langkah 3 & 4)
targets = {
    "Awal": 2,
    "Tengah": 35,
    "Akhir": 85,
    "Tidak Ada": 90
}

print("=== TRACING PENYEMPITAN RUANG PENCARIAN BINARY SEARCH ===")
comp_results = {
    'Skenario': [],
    'Target': [],
    'Indeks': [],
    'Langkah Binary Search': [],
    'Langkah Linear Search': []
}

for label, target in targets.items():
    print(f"\n--- Menguji Target ({label}): {target} ---")
    idx_bin, steps_bin = binary_search_recursive(data_terurut, target, 0, len(data_terurut) - 1)
    idx_lin, steps_lin = linear_search(data_terurut, target)

    comp_results['Skenario'].append(label)
    comp_results['Target'].append(target)
    comp_results['Indeks'].append(idx_bin)
    comp_results['Langkah Binary Search'].append(steps_bin)
    comp_results['Langkah Linear Search'].append(steps_lin)

# 4. Tampilkan Tabel Perbandingan Langkah Menggunakan Pandas
print("\n" + "="*50)
print("=== TABEL PERBANDINGAN JUMLAH LANGKAH ===")
df_comp = pd.DataFrame(comp_results)
df_comp.style.hide(axis="index")
```
