Contoh 1.4 — Fungsi Pencarian Linear Sederhana (Modifikasi)

```python
def cari(data, target):
    for i in range(len(data)):
        if data[i] == target:
            return i
    return -1
```

```python
data_awal = [12, 20, 7, 30]
print("--- Contoh Awal ---")
print("Cari 7 pada data_awal:", cari(data_awal, 7))   # Output: 2
print("Cari 99 pada data_awal:", cari(data_awal, 99)) # Output: -1
```

```python
data_baru = [50, 15, 88, 42, 100, 23]
target1 = 88
target2 = 50
target3 = 999

print("\n--- Modifikasi Data & Target ---")
print("Data baru:", data_baru)
print(f"Indeks target ({target1}):", cari(data_baru, target1)) # Ditemukan di indeks 2
print(f"Indeks target ({target2}):", cari(data_baru, target2)) # Ditemukan di indeks 0
print(f"Indeks target ({target3}):", cari(data_baru, target3)) # Tidak ditemukan -> -1
```

Contoh 1.5 — Bentuk Awal Node (Modifikasi)

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
node_A = Node(100)
node_B = Node("Data Struktur")
```

```python
print("--- Objek Node A ---")
print("Data Node A:", node_A.data)
print("Next Node A:", node_A.next)

print("\n--- Objek Node B ---")
print("Data Node B:", node_B.data)
print("Next Node B:", node_B.next)
```
