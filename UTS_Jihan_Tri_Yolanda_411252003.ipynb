```python
# ==========================================
# 1. PERSIAPAN DATA AWAL
# ==========================================

# Ganti nilai d dengan digit terakhir NIM Anda
d = 7

# Data mahasiswa awal
data_awal = [
    ['TI001', 'Alya', 78],
    ['TI002', 'Bima', 80 + d],
    ['TI003', 'Citra', 67],
    ['TI004', 'Danu', 92],
    ['TI005', 'Eka', 74],
    ['TI006', 'Fajar', 88],
    ['TI007', 'Gina', 81],
    ['TI008', 'Hadi', 65 + d],
    ['TI009', 'Intan', 91],
    ['TI010', 'Joko', 76],
    ['TI011', 'Karin', 84],
    ['TI012', 'Lukman', 70 + d]
]

# 1. Fungsi Linear Search Manual
def linear_search_nim(data, target):
    comparisons = 0
    for i in range(len(data)):
        comparisons += 1
        if data[i][0] == target:
            return i, comparisons  # Mengembalikan indeks hasil dan jumlah perbandingan
    return -1, comparisons

# 2. Penentuan Target Sesuai Rumus
target_index = d % len(data_awal)
target_nim = data_awal[target_index][0]

print("=== PENGUJIAN LINEAR SEARCH (DATA SEBELUM DIURUTKAN) ===")
print(f"Target NIM (Ada)      : {target_nim} (pada indeks acuan {target_index})")

idx_found, comp_found = linear_search_nim(data_awal, target_nim)
print(f"Hasil Cari '{target_nim}' : Indeks = {idx_found}, Jumlah Perbandingan = {comp_found}")

idx_not_found, comp_not_found = linear_search_nim(data_awal, "TI999")
print(f"Hasil Cari 'TI999'    : Indeks = {idx_not_found}, Jumlah Perbandingan = {comp_not_found}")
print("-" * 60)


# 3. Fungsi Merge dan Merge Sort Rekursif Manual
def merge(left, right):
    result = []
    i = 0
    j = 0

    # Membandingkan NIM elemen left dan right secara manual
    while i < len(left) and j < len(right):
        if left[i][0] <= right[j][0]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    # Menggabungkan sisa elemen jika ada
    result.extend(left[i:])
    result.extend(right[j:])
    return result

def merge_sort_nim(data):
    # Base case rekursi
    if len(data) <= 1:
        return data

    mid = len(data) // 2
    left_half = merge_sort_nim(data[:mid])
    right_half = merge_sort_nim(data[mid:])

    return merge(left_half, right_half)


# 4. Tampilkan Data Sebelum dan Sesudah Pengurutan
# Menggunakan salinan data agar tidak mengubah data_awal secara tidak sengaja
data_untuk_diurutkan = [row[:] for row in data_awal]
data_terurut = merge_sort_nim(data_untuk_diurutkan)

print("\n=== DATA SEBELUM PENGURUTAN ===")
for mhs in data_awal:
    print(mhs)

print("\n=== DATA SESUDAH PENGURUTAN (MERGE SORT) ===")
for mhs in data_terurut:
    print(mhs)
print("-" * 60)


# 5. Fungsi Binary Search Manual
def binary_search_nim(data_urut, target):
    low = 0
    high = len(data_urut) - 1
    comparisons = 0

    while low <= high:
        comparisons += 1
        mid = (low + high) // 2

        if data_urut[mid][0] == target:
            return mid, comparisons
        elif data_urut[mid][0] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1, comparisons


# 6. Pengujian Binary Search
print("\n=== PENGUJIAN BINARY SEARCH (DATA TERURUT) ===")
bs_idx_found, bs_comp_found = binary_search_nim(data_terurut, target_nim)
print(f"Hasil Cari '{target_nim}' : Indeks = {bs_idx_found}, Jumlah Perbandingan = {bs_comp_found}")

bs_idx_not, bs_comp_not = binary_search_nim(data_terurut, "TI999")
print(f"Hasil Cari 'TI999'    : Indeks = {bs_idx_not}, Jumlah Perbandingan = {bs_comp_not}")
print("-" * 60)


# 8. Ringkasan Hasil Pengerjaan
print("\n=== RINGKASAN HASIL PERBANDINGAN PENCARIAN ===")
print(f"Target DICARI                 : {target_nim}")
print(f"Indeks Hasil Linear Search    : {idx_found}")
print(f"Indeks Hasil Binary Search    : {bs_idx_found}")
print(f"Jumlah Perbandingan Linear    : {comp_found}")
print(f"Jumlah Perbandingan Binary    : {bs_comp_found}")
```

## 9. Analisis Kompleksitas & Teori Algoritma

**1. Kompleksitas Waktu (Time Complexity)**
* **Linear Search:**
  * *Best Case:* $\mathcal{O}(1)$ — terjadi jika data yang dicari berada di posisi paling awal (indeks pertama).
  * *Worst / Average Case:* $\mathcal{O}(n)$ — terjadi jika data berada di posisi paling akhir atau tidak ditemukan sama sekali, sehingga algoritma harus memeriksa seluruh elemen satu per satu sebanyak $n$ kali.
* **Binary Search:**
  * *Best Case:* $\mathcal{O}(1)$ — terjadi jika elemen tengah (`mid`) pada iterasi pertama langsung cocok dengan data yang dicari.
  * *Worst / Average Case:* $\mathcal{O}(\log n)$ — terjadi karena ruang pencarian terus dibagi dua (halving) pada setiap langkah hingga data ditemukan atau batas pencarian habis.
* **Merge Sort:**
  * *Best, Average, & Worst Case:* $\mathcal{O}(n \log n)$ — proses pembagian data secara rekursif (*divide*) membutuhkan waktu sebesar $\mathcal{O}(\log n)$ level, dan pada setiap level dibutuhkan proses penggabungan (*conquer/merge*) linier sebesar $\mathcal{O}(n)$.

---

**2. Kebutuhan Ruang Tambahan (Space Complexity) Merge Sort**

Merge Sort memiliki kompleksitas ruang tambahan sebesar $\mathcal{O}(n)$.
* **Alasan:** Merge Sort bukan algoritma pengurutan di tempat (*in-place sorting*). Saat melakukan pembagian data secara rekursif dan penggabungan (*merge*), algoritma membuat salinan *array/list* baru (seperti variabel `left` dan `right`) untuk menampung elemen sementara sebelum digabungkan kembali dalam keadaan terurut.

---

**3. Alasan Binary Search Memerlukan Data Terurut**

Binary Search bekerja menggunakan prinsip **eliminasi ruang pencarian**. Algoritma ini membandingkan target dengan elemen tengah (`mid`).
* Jika data **terurut**, keputusan untuk mengeliminasi setengah bagian data (kiri atau kanan) dipastikan aman dan valid.
* Jika data **tidak terurut (acak)**, perbandingan dengan elemen tengah tidak memberikan petunjuk logis apakah target berada di sebelah kiri atau kanan `mid`, sehingga Binary Search akan gagal menemukan elemen yang dicari.

---

**4. Kapan Biaya Sorting Layak Dibayar Sebelum Pencarian?**

Biaya *sorting* sebesar $\mathcal{O}(n \log n)$ **layak dibayar** ketika aplikasi melakukan **pencarian berulang kali (multiple/frequent searches)** pada himpunan data yang sama dan jarang berubah.

* **Satu kali pencarian ($k = 1$):**
  Menggunakan Linear Search lebih efisien ($\mathcal{O}(n)$) daripada mengurutkan data terlebih dahulu baru melakukan Binary Search ($\mathcal{O}(n \log n) + \mathcal{O}(\log n)$).
* **Pencarian berulang ($k$ kali pencarian):**
  Jika dilakukan $k$ kali pencarian di mana $k$ cukup besar, total biaya pencarian:
  $$\text{Linear Search} = \mathcal{O}(k \cdot n)$$
  $$\text{Sort + Binary Search} = \mathcal{O}(n \log n + k \log n)$$
  Karena $\log n \ll n$, untuk nilai $k$ yang besar, biaya pengurutan di awal menjadi sangat murah dibanding harus melakukan Linear Search berulang kali.

# SOAL 2: Sistem Antrean Layanan dengan Undo Menggunakan Linked Queue dan Stack

```python
# ==========================================
# 1. IMPLEMENTASI STRUKTUR DATA UTAMA
# ==========================================

# Class Node untuk Linked List
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# Class LinkedQueue berbasis Singly Linked List
class LinkedQueue:
    def __init__(self):
        self.front = None
        self.rear = None
        self.count = 0

    def is_empty(self):
        return self.front is None

    def size(self):
        return self.count

    def enqueue(self, data):
        new_node = Node(data)
        if self.is_empty():
            self.front = new_node
            self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node
        self.count += 1

    def enqueue_front(self, data):
        """Menambahkan elemen ke bagian paling depan Queue (digunakan saat UNDO)"""
        new_node = Node(data)
        if self.is_empty():
            self.front = new_node
            self.rear = new_node
        else:
            new_node.next = self.front
            self.front = new_node
        self.count += 1

    def dequeue(self):
        if self.is_empty():
            print("  [Peringatan] Queue Kosong! Tidak ada antrean yang bisa dilayani.")
            return None

        removed_data = self.front.data
        self.front = self.front.next
        self.count -= 1

        if self.front is None:
            self.rear = None

        return removed_data

    def peek(self):
        if self.is_empty():
            print("  [Peringatan] Queue Kosong!")
            return None
        return self.front.data

    def display(self):
        if self.is_empty():
            return "[] (Kosong)"
        elements = []
        curr = self.front
        while curr:
            elements.append(str(curr.data))
            curr = curr.next
        return " -> ".join(elements) + " (Front -> Rear)"


# Class Stack berbasis Python List (Wrapper)
class Stack:
    def __init__(self):
        self._items = []

    def is_empty(self):
        return len(self._items) == 0

    def size(self):
        return len(self._items)

    def push(self, data):
        self._items.append(data)

    def pop(self):
        if self.is_empty():
            print("  [Peringatan] Stack Kosong! Tidak ada riwayat untuk di-UNDO.")
            return None
        return self._items.pop()

    def peek(self):
        if self.is_empty():
            print("  [Peringatan] Stack Kosong!")
            return None
        return self._items[-1]

    def display(self):
        if self.is_empty():
            return "[] (Kosong)"
        return " -> ".join(map(str, reversed(self._items))) + " (Top -> Bottom)"


# ==========================================
# 2. PENGUJIAN UJI COBA OPERASI DASAR CLASS
# ==========================================

print("=== PENGUJIAN OPERASI DASAR QUEUE DAN STACK ===")
test_q = LinkedQueue()
test_s = Stack()

print("Status awal Queue kosong:", test_q.is_empty())
print("Status awal Stack kosong:", test_s.is_empty())

test_q.enqueue("TEST-01")
test_q.enqueue("TEST-02")
print("Queue setelah Enqueue 2 elemen:", test_q.display())
print("Dequeue elemen:", test_q.dequeue())
print("Queue setelah Dequeue:", test_q.display())

test_s.push("TEST-01")
print("Stack setelah Push 1 elemen:", test_s.display())
print("Pop elemen Stack:", test_s.pop())
print("Stack setelah Pop:", test_s.display())
print("-" * 65)


# ==========================================
# 3. SIMULASI UTAMA (12 OPERASI)
# ==========================================

d = 7  # Ganti dengan digit terakhir NIM Anda

# Inisialisasi ID Layanan
A = f'MHS-{d}01'
B = f'MHS-{d}02'
C = f'MHS-{d}03'
D = f'MHS-{d}04'
E = f'MHS-{d}05'
F = f'MHS-{d}06'

queue = LinkedQueue()
stack_history = Stack()

operations = [
    ("DATANG A", A),
    ("DATANG B", B),
    ("DATANG C", C),
    ("LAYANI", None),
    ("DATANG D", D),
    ("LAYANI", None),
    ("DATANG E", E),
    ("UNDO", None),
    ("LAYANI", None),
    ("LAYANI", None),
    ("UNDO", None),
    ("DATANG F", F)
]

print("\n=== JALANNYA SIMULASI ANTREAN LAYANAN MAHASISWA ===")

for step, (op_name, val) in enumerate(operations, start=1):
    print(f"\nLangkah {step}: {op_name}")

    if op_name.startswith("DATANG"):
        queue.enqueue(val)
        print(f"  --> {val} masuk ke antrean.")

    elif op_name == "LAYANI":
        served = queue.dequeue()
        if served:
            stack_history.push(served)
            print(f"  --> {served} selesai dilayani & masuk riwayat.")

    elif op_name == "UNDO":
        last_served = stack_history.pop()
        if last_served:
            queue.enqueue_front(last_served)
            print(f"  --> Membatalkan pelayanan {last_served}, dikembalikan ke Depan Queue.")

    print(f"  Queue : {queue.display()}")
    print(f"  Stack : {stack_history.display()}")

print("\n" + "=" * 65)
print("                       RINGKASAN AKHIR SIMULASI                      ")
print("=" * 65)
front_val = queue.peek() if not queue.is_empty() else "Kosong"
rear_val = queue.rear.data if queue.rear else "Kosong"

print(f"Front Queue Saat Ini : {front_val}")
print(f"Rear Queue Saat Ini  : {rear_val}")
print(f"Jumlah Elemen Queue  : {queue.size()}")
print(f"Jumlah Elemen Stack  : {stack_history.size()}")
print("=" * 65)


# ==========================================
# 4. PENGUJIAN EDGE CASE (PENANGANAN ERROR)
# ==========================================

print("\n=== PENGUJIAN EDGE CASE (PENANGANAN KONDISI KOSONG) ===")

empty_q = LinkedQueue()
empty_s = Stack()

print("1. Menguji Dequeue pada Queue Kosong:")
res_q = empty_q.dequeue()
print("   Hasil Dequeue:", res_q)

print("\n2. Menguji Pop / UNDO pada Stack Kosong:")
res_s = empty_s.pop()
print("   Hasil Pop:", res_s)
print("\nProgram berhasil menangani edge case tanpa crash!")
```

## 5. Analisis Prinsip FIFO (Queue) dan LIFO (Stack) dari Simulasi

1. **Prinsip FIFO (First-In, First-Out) pada Queue:**
   * **Penjelasan:** Elemen yang pertama kali masuk ke dalam antrean akan menjadi elemen pertama yang keluar/dilayani.
   * **Bukti Simulasi:** Pada langkah 1-3, `MHS-701` masuk pertama, disusul `MHS-702` dan `MHS-703`. Ketika operasi `LAYANI` dijalankan di Langkah 4, elemen yang diproses adalah `MHS-701` (elemen paling depan), sesuai dengan aturan antrean FIFO[cite: 4].

2. **Prinsip LIFO (Last-In, First-Out) pada Stack:**
   * **Penjelasan:** Elemen yang terakhir kali dimasukkan/diproses adalah yang pertama kali dikeluarkan.
   * **Bukti Simulasi:** Pada langkah 4 `MHS-701` dilayani (masuk Stack), lalu langkah 6 `MHS-702` dilayani (masuk Stack). Ketika dilakukan `UNDO` pada Langkah 8, ID yang ditarik dari Stack riwayat adalah `MHS-702` karena ia merupakan elemen paling atas (terakhir dilayani)[cite: 4].

## 6. Analisis Kompleksitas Waktu & Efisiensi Pointer

1. Tabel Kompleksitas Waktu Operasi

| Class / Operasi | Method | Kompleksitas Waktu | Keterangan |
| :--- | :--- | :--- | :--- |
| **LinkedQueue** | `enqueue(data)` | $\mathcal{O}(1)$ | Langsung ditambahkan via pointer `rear` |
| | `enqueue_front(data)` | $\mathcal{O}(1)$ | Langsung disisipkan via pointer `front` |
| | `dequeue()` | $\mathcal{O}(1)$ | Langsung dihapus via pointer `front` |
| | `peek()` | $\mathcal{O}(1)$ | Membaca atribut `front.data` |
| **Stack** | `push(data)` | $\mathcal{O}(1)$ *(Amortized)* | Menambahkan ke ujung Python List |
| | `pop()` | $\mathcal{O}(1)$ | Menghapus elemen dari ujung Python List |
| | `peek()` | $\mathcal{O}(1)$ | Membaca elemen terakhir List |

2. Mengapa Pointer `front` dan `rear` Membuat Queue Efisien?
Dengan menyimpan referensi pointer langsung ke node awal (`front`) dan node akhir (`rear`), kita tidak perlu melakukan penelusuran (*traversal*) dari awal hingga ujung linked list saat menambah data baru. Operasi penambahan di belakang dan penghapusan di depan cukup memperbarui nilai referensi pointer terkait, sehingga kompleksitasnya konstan $\mathcal{O}(1)$[cite: 5].

## 7. Perbandingan Linked List Queue vs Python List `pop(0)`

Menggunakan Linked List jauh lebih tepat dibanding `list.pop(0)` untuk antrean yang sering melakukan operasi *dequeue*:

1. **Operasi `list.pop(0)` pada Python List:**
   * Memiliki kompleksitas waktu **$\mathcal{O}(n)$**.
   * **Alasan:** Ketika elemen paling depan (`index 0`) dihapus dari array/list kontigu, Python harus menggeser seluruh $n-1$ elemen sisa satu posisi ke kiri di dalam memori. Semakin banyak data antrean, kinerjanya akan semakin lambat secara drastis[cite: 5].

2. **Operasi `dequeue()` pada Linked List Queue:**
   * Memiliki kompleksitas waktu **$\mathcal{O}(1)$**.
   * **Alasan:** Penghapusan dilakukan hanya dengan menggeser pointer `front` ke node berikutnya (`self.front = self.front.next`). Tidak ada elemen lain yang perlu digeser di memori, sehingga operasi selalu konstan terlepas dari berapa pun banyaknya jumlah antrean[cite: 5].
